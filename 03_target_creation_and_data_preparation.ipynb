```python
# 1. Import Libraries

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler

import warnings
warnings.filterwarnings('ignore')

print("✅ Libraries imported successfully")

```

```python
# 2. Load Feature Engineered Data

train_df = pd.read_csv('../data/train_featured.csv')
test_df  = pd.read_csv('../data/test_featured.csv')

print("Train Shape:", train_df.shape)
print("Test Shape :", test_df.shape)
```

```python
# 3. Target Variable Creation 
# Classification Target – Customer Conversion

train_df['converted'] = np.where(
    (train_df['page'] > 1) & (train_df['price_2'] == 1),
    1, 0
)

test_df['converted'] = np.where(
    (test_df['page'] > 1) & (test_df['price_2'] == 1),
    1, 0
)

print("✅ Classification target created")
print(train_df['converted'].value_counts())
```

```python
# Regression Target – Revenue Estimation

train_df['revenue'] = train_df['price'] * train_df['total_clicks']
test_df['revenue']  = test_df['price'] * test_df['total_clicks']

print("✅ Regression target created")
```

```python
# 4. Drop Non- Informative Columns

drop_cols = ['session_id', 'page2_clothing_model']

train_df.drop(columns=drop_cols, inplace=True)
test_df.drop(columns=drop_cols, inplace=True)

print("⚠️ Dropped ID / high-cardinality columns:", drop_cols)
```

```python
# 5. Categorical Encoding

cat_cols = train_df.select_dtypes(include='object').columns
le = LabelEncoder()

for col in cat_cols:
    train_df[col] = le.fit_transform(train_df[col])
    test_df[col]  = le.transform(test_df[col])

print("✅ Categorical features encoded")
```

```python
# 6. Feature-Target Split
# Classification

X_cls = train_df.drop(['converted','revenue'], axis=1)
y_cls = train_df['converted']
```

```python
# Regression

X_reg = train_df.drop(['converted','revenue'], axis=1)
y_reg = train_df['revenue']
```

```python
# Clustering

X_cluster = train_df.drop(['converted','revenue'], axis=1)
```

```python
print("✅ Features and targets separated")
```

```python
# 7. Feature Scaling

scaler = StandardScaler()

X_cls_scaled = scaler.fit_transform(X_cls)
X_reg_scaled = scaler.fit_transform(X_reg)
X_cluster_scaled = scaler.fit_transform(X_cluster)

print("⚠️ Features scaled using StandardScaler")
```

```python
# 8. Class Imbalance Check 

class_ratio = y_cls.value_counts(normalize=True)

print("⚠️ Class Distribution:")
print(class_ratio)

if class_ratio.min() < 0.3:
    print("⚠️ Imbalance detected → SMOTE will be applied in Step 4")
else:
    print("✅ Classes reasonably balanced")
```

```python
# 9. Save Model-Ready Datasets 

pd.DataFrame(X_cls_scaled).to_csv('../data/X_classification.csv', index=False)
pd.DataFrame(X_reg_scaled).to_csv('../data/X_regression.csv', index=False)
pd.DataFrame(X_cluster_scaled).to_csv('../data/X_clustering.csv', index=False)

y_cls.to_csv('../data/y_classification.csv', index=False)
y_reg.to_csv('../data/y_regression.csv', index=False)

print("✅ Step 3 completed successfully")
print("📁 Model-ready datasets saved in data folder")
```
