```python
# STEP 5.1 – Import Libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingRegressor

from sklearn.metrics import (
    accuracy_score, classification_report,
    mean_squared_error, r2_score
)

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

import joblib
import warnings
warnings.filterwarnings('ignore')

print("✅ Libraries imported")
```

```python
# STEP 5.2 – Load Model-Ready Data

X_cls = pd.read_csv('../data/X_classification.csv')
y_cls = pd.read_csv('../data/y_classification.csv')

X_reg = pd.read_csv('../data/X_regression.csv')
y_reg = pd.read_csv('../data/y_regression.csv')

X_cluster = pd.read_csv('../data/X_clustering.csv')

print("✅ Model-ready data loaded")
```

```python
# STEP 5.3 – Train–Test Split
# Classification split

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls, y_cls,
    test_size=0.2,
    random_state=42,
    stratify=y_cls
)
```

```python
# Regression split
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg,
    test_size=0.2,
    random_state=42
)
```

```python
print("✅ Train-test split completed")
```

```python
# STEP 5.4 – Classification Evaluation

rf = joblib.load('../models/classification_model.pkl')

y_pred = rf.predict(X_test_cls)

print("Accuracy:", accuracy_score(y_test_cls, y_pred))
print(classification_report(y_test_cls, y_pred))
```

```python
# STEP 5.5 – Regression Evaluation

gbr = joblib.load('../models/regression_model.pkl')

y_pred_reg = gbr.predict(X_test_reg)

print("RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred_reg)))
print("R2 Score:", r2_score(y_test_reg, y_pred_reg))
```

```python
# STEP 5.6 – Clustering Evaluation

kmeans = joblib.load('../models/clustering_model.pkl')

clusters = kmeans.predict(X_cluster)

print("Silhouette Score:",
      silhouette_score(X_cluster, clusters))
```

```python
# STEP 5.7 – Hyperparameter Tuning

param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [10, 15]
}

grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid,
    cv=3,
    scoring='f1',
    n_jobs=-1
)

grid.fit(X_train_cls, y_train_cls)

print("Best Params:", grid.best_params_)
```

```python
# STEP 5.8 – Save Tuned Model

joblib.dump(grid.best_estimator_,
            '../models/classification_model_tuned.pkl')

print("✅ Tuned model saved")
```
