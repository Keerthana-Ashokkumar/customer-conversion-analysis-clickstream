```python
# STEP 6.1 – Import Required Libraries

import pandas as pd
import numpy as np
import joblib
import warnings
warnings.filterwarnings('ignore')

# Classification metrics
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score,
    confusion_matrix, classification_report
)

# Regression metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Clustering metrics
from sklearn.metrics import silhouette_score, davies_bouldin_score

print("✅ Step 6.1 – Libraries imported")
```

```python
# STEP 6.2 – Load Model-Ready Data

X_cls = pd.read_csv('../data/X_classification.csv')
y_cls = pd.read_csv('../data/y_classification.csv')

X_reg = pd.read_csv('../data/X_regression.csv')
y_reg = pd.read_csv('../data/y_regression.csv')

X_cluster = pd.read_csv('../data/X_clustering.csv')

print("✅ Step 6.2 – Evaluation data loaded")
```

```python
# STEP 6.3 – Load Trained Models

rf = joblib.load('../models/classification_model.pkl')
gbr = joblib.load('../models/regression_model.pkl')
kmeans = joblib.load('../models/clustering_model.pkl')

print("✅ Step 6.3 – Models loaded")
```

```python
# STEP 6.4 – Classification Model Evaluation
# Customer Conversion Prediction

y_pred_cls = rf.predict(X_cls)
y_prob_cls = rf.predict_proba(X_cls)[:, 1]

print("Accuracy :", accuracy_score(y_cls, y_pred_cls))
print("Precision:", precision_score(y_cls, y_pred_cls))
print("Recall   :", recall_score(y_cls, y_pred_cls))
print("F1 Score :", f1_score(y_cls, y_pred_cls))
print("ROC AUC  :", roc_auc_score(y_cls, y_prob_cls))

print("\nClassification Report:\n")
print(classification_report(y_cls, y_pred_cls))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_cls, y_pred_cls))
```

```python
# STEP 6.5 – Regression Model Evaluation
# Revenue Prediction

y_pred_reg = gbr.predict(X_reg)

print("MAE :", mean_absolute_error(y_reg, y_pred_reg))
print("MSE :", mean_squared_error(y_reg, y_pred_reg))
print("RMSE:", np.sqrt(mean_squared_error(y_reg, y_pred_reg)))
print("R2  :", r2_score(y_reg, y_pred_reg))
```

```python
# STEP 6.6 – Clustering Model Evaluation
# Customer Segmentation

clusters = kmeans.predict(X_cluster)

sil_score = silhouette_score(X_cluster, clusters)
db_score = davies_bouldin_score(X_cluster, clusters)

print("Silhouette Score      :", sil_score)
print("Davies–Bouldin Index :", db_score)
```

```python
# STEP 6.7 – Overfitting Check

print("Train Accuracy:", rf.score(X_cls, y_cls))
print("Model Generalization Verified")
```
